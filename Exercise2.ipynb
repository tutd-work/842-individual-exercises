```python
import hashlib
import json
import pickle
import statistics
import time

# Initialize a block
block = {
    "index": 3,
    "timestamp": 1787472618,
    "previous_hash": "1a908156af6faac807ffb91eda455e5adac1ec9bb371ee382b19837f7400d23c",
    "transactions": [
        "Transfer 5 coins from user6 to user9",
        "Transfer 4 coins from user3 to user5"
    ],
    "merkle_root": "dd8e0934f1e9c9e1c40718b8918d209464078e843c198b7b7be8e6f4d42a27a4",
    "hash": "662b260a8c283d15db448d3848d34e020620c1f1e191f523197de2d57a7aea72"
}

# Serialize the block using pickle
pickled_block = pickle.dumps(block)
```

```python
# Initialize mining parameters
nonce = 1
target_blocks = 30

# Initialize variables for mining statistics
hash_attempts = 0
mined_blocks = 0
valid_hashs = []
mine_times = []

# Mining 30 blocks with a difficulty level
difficulty = 5
target = "0" * difficulty
while mined_blocks < target_blocks:
    start_time = time.time()
    block_found = False
    while not block_found:
        working_block = hashlib.sha256(str(nonce).encode() + pickled_block).hexdigest()
        hash_attempts += 1

        # Check if the hash meets the target difficulty
        if working_block.startswith(target):
            block_found = True
            mined_blocks += 1
            valid_hashs.append(working_block)
        nonce += 1

    end_time = time.time()
    current_time = end_time - start_time

    # Update time statistics
    mine_times.append(current_time)
print(f"Difficulty level: {difficulty} - Total time: {sum(mine_times):,.5f} seconds ")
print(f"Average time: {sum(mine_times) / len(mine_times):,.5f} seconds")
print(f"Minimum time: {min(mine_times):,.5f} seconds")
print(f"Maximum time: {max(mine_times):,.5f} seconds")
print(f"Standard deviation: {(statistics.stdev(mine_times)):,.5f} seconds")
print(f"Total hash attempts: {hash_attempts:,}")
print(f"Mining throughput: {hash_attempts/sum(mine_times):,.2f} hashes/second")
print(f"Theoretical hash attempts: {16**len(target):,}")
print(f"Valid hashes: {json.dumps(valid_hashs, indent=2)}")
```
