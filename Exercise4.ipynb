```python
def calculate_bitcoin_tps(max_block_size):
    average_tx_size = 580  # bytes
    block_time = 576       # seconds

    transactions_per_block = max_block_size / average_tx_size
    tps = transactions_per_block / block_time

    return transactions_per_block, tps

# Calculate the theoretical transaction throughput
for block_size_mb in [1, 2, 4]:
    block_size = block_size_mb * 1_000_000
    tx_per_block,tps = calculate_bitcoin_tps(block_size)

    print(f"Block size: {block_size_mb} MB, " +
          f"Transactions per block: {tx_per_block:,.2f}, TPS: {tps:.2f}")
```
